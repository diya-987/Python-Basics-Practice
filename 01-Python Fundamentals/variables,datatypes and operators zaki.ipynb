##### NAMING RULES

1.Should be start with an alphabet or underscore

```python
a=10
print(a)

```

```python
_b=39
print(_b)
```

2.Don't start with a number

```python
12number=4
print(12number)
```

```python
_12number=4
print(_12number)
```

3.Case sensitivity

```python
age=27
print(Age)
```

```python
age=27
print(age)
```

```python
AGE=27
print(Age)
```

```python
AGE=27
print(AGE)
```

4.Should only contain alphanumerics and underscore

```python
abc_123=46
print(abc_123)
```

```python
abc-123!>=65
print(abc-123!>)
```
