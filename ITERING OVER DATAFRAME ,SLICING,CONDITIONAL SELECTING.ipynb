## ITERATION OVER DATAFRAME:

```python
*Its used when you want to loop through rows or columns.
```

```python
#1.iterate over columns:
import pandas as pd
df=pd.DataFrame({
    'name':['A','B','C'],
    'age':[20,25,30]
})
print(df)

for col in df:
    print(col)
```

```python
#2.iterate over rows:
for index,row in df.iterrows():
    print(index,row)
```

```python
#3.iterating using itertuples()
#its faster than iterrows()
for row in df.itertuples():
    print(row)
```

```python
#access using dot(row.name)
for row in df.itertuples():
    print(row.name,row.age)
```

```python
#4.iterating using items():(column wise)
for col_name,col_values in df.items():
    print(col_name)
    print(col_values)
```

### CONDITIONAL SELECTION(FILTER):

```python
Used to select data based on conditions.
```

```python
#1.single condition:
df[df['age']>20]
```

```python
#2.multiple conditions(&):
df[(df['name']=='C')&(df['age']>20)]
```

```python
df[(df['name']=='A')&(df['age']==20)]
```

```python
#3.multiple conditions with or(|):
df[(df['age']==25)|(df['name']=='C')]
```
