## 1.LIST:

```python
A list is a collection data type used to store multiple values
in one variable.
```

```python
*values are stored in a square bracket-->[]
*ordered-->elements have a fixed possition(index)
*mutable-->elements  can be changed
*allow duplicates-->[1,1,4,7,7]
*can store different data types-->[1,2,'fathima',True,9.5]
```

```python
LIST FUNCTIONS:
1.append()
2.extend()
3.insert()
4.remove()
5.pop()
6.clear()
7.len()
8.max()
9.min()
10.sum()
11.sort()
12.reverse()
13.variable[::-1]----to reverse in descending order
14.count()-->to count the occurance of an element
15.index()-->to find the possition of an element
```

# 3.SET

```python
a set is a collection data tyupe used to store
multiple values like list and tuple,but without duplicates.
```

```python
features:
*unordered
*no indexing
*mutable
*does not allow duplicates
*stores unique elements only
```

```python
a={"a","b","c"}
a
```

```python
#remove repeated elements from a list by coverting into set.
a=[1,2,2,3,3]
b=set(a)
c=list(b)
c
```

```python
#integer set
a={1,2,5,8,9,4}
print(a)
#string set
b={"a","b","c"}
print(b)
#float set
c={2.5,6.2,8.7}
print(c)
#multiple set
d={1,2,"b","c",6.2,8.7}
print(d)
```
