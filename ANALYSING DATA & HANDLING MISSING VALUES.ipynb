### HANDLING MISSING VALUES:

```python
#fill missing values by using fillna()
```

```python
#1.fill with constand value:
df.fillna('A')
```

```python
#2.fill with mean
avg=df['AGE'].mean()
df['AGE']=df['AGE'].fillna(avg)
df
```

```python
#3.fill with median
avg=df['SALARY'].median()
df['SALARY']=df['SALARY'].fillna(avg)
df
```

```python
#4.fill with forward fill(ffill)
df.fillna(method='ffill')
```

```python
#5.fill with backward fill(bfill)
df.fillna(method='bfill')
```

```python
#6.by using replace
df.replace(np.nan,'a')
```

```python
#7.drop rows
df.dropna(axis=0)
```

```python
#8.drop columns
df.dropna(axis=1)
```

```python
#to drop column
df.drop('CITY',axis=1)
```
